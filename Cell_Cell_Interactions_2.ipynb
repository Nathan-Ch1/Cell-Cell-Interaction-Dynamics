```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import torch
from scipy.signal import savgol_filter
```

```python
np.random.seed(42)


# load data

data = np.genfromtxt("pnas.2016602118.sd02.txt")
print("Raw data shape:", data.shape)


# parameters

dt = 10.0
num_cells = data.shape[0]

# small constant to prevent division by zero

epsilon = 0.05

# Gaussian basis parameters

centers = np.linspace(0, 1, 5)
sigma = 0.15


# shared interaction feature library

# this SAME function is used during both
# training and trajectory prediction
def build_phi_library(R_scaled):

    gaussians = [
        np.exp(
            -(R_scaled - c)**2 /
            (2 * sigma**2)
        )
        for c in centers
    ]

    phi_library = np.column_stack([
        np.ones_like(R_scaled),
        R_scaled,
        1.0 / (R_scaled + epsilon),
        1.0 / (R_scaled + epsilon)**2,
        np.exp(-R_scaled),
        np.exp(-2 * R_scaled),
        *gaussians
    ])

    return phi_library


# cells are already paired:

# (0,1)

# (2,3)

# (4,5)

# make pairs

pairs = []

for i in range(0, num_cells - 1, 2):
    pairs.append((i, i + 1))

print("Number of interacting pairs:", len(pairs))


# choose ONE interacting pair

pair_index = 23
i, j = pairs[pair_index]

print("Using pair:", (i, j))


# trajectory of cell i
xi = data[i]


# trajectory for cell j
xj = data[j]


# remove NaNs
xi = xi[~np.isnan(xi)]
xj = xj[~np.isnan(xj)]


# get shorter length, make both equal
length = min(len(xi), len(xj))

# need enough points for Savitzky-Golay filtering
if length < 21:
    raise ValueError("Not enough trajectory points for Savitzky-Golay filtering.")

# fix lengths to shorter one so both are same length
xi = xi[:length]
xj = xj[:length]


# smooth positions
xi_smooth = savgol_filter(
    xi,
    window_length=21,
    polyorder=3
)

xj_smooth = savgol_filter(
    xj,
    window_length=21,
    polyorder=3
)


# calculate velocity directly using Savitzky-Golay derivative
vi = savgol_filter(
    xi,
    window_length=21,
    polyorder=3,
    deriv=1,
    delta=dt
)

vj = savgol_filter(
    xj,
    window_length=21,
    polyorder=3,
    deriv=1,
    delta=dt
)


# calculate acceleration directly using Savitzky-Golay derivative
vdot_i = savgol_filter(
    xi,
    window_length=21,
    polyorder=3,
    deriv=2,
    delta=dt
)

vdot_j = savgol_filter(
    xj,
    window_length=21,
    polyorder=3,
    deriv=2,
    delta=dt
)


# use smoothed positions
xi = xi_smooth
xj = xj_smooth

# get distance
r = np.abs(xi - xj)


# split trajectory into first 80% and final 20%

split = int(0.8 * len(r))
print("Training trajectory points:", split)
print("Testing trajectory points:", len(r) - split)


# split positions, velocities, distances, and acceleration
xi_train = xi[:split]
xi_test = xi[split:]

xj_train = xj[:split]
xj_test = xj[split:]

vi_train = vi[:split]
vi_test = vi[split:]

vj_train = vj[:split]
vj_test = vj[split:]

vdot_i_train = vdot_i[:split]
vdot_i_test = vdot_i[split:]

vdot_j_train = vdot_j[:split]
vdot_j_test = vdot_j[split:]

r_train = r[:split]
r_test = r[split:]


# normalize distance for this pair
max_r = np.max(r_train)

if max_r == 0:
    raise ValueError("Maximum training distance is zero.")

R_train_scaled = r_train / max_r
R_test_scaled = r_test / max_r


# interaction library

# this is the SAME library used later
# during trajectory prediction
phi_library_train = build_phi_library(R_train_scaled)
phi_library_test = build_phi_library(R_test_scaled)


# SINDy library for F(x, v) cell i
F_i_train = np.column_stack([
    np.ones_like(xi_train),
    xi_train,
    xi_train**2,
    xi_train**3,
    vi_train,
    vi_train**2,
    xi_train * vi_train
])


F_i_test = np.column_stack([
    np.ones_like(xi_test),
    xi_test,
    xi_test**2,
    xi_test**3,
    vi_test,
    vi_test**2,
    xi_test * vi_test
])


# SINDy library for F(x, v) cell j
F_j_train = np.column_stack([
    np.ones_like(xj_train),
    xj_train,
    xj_train**2,
    xj_train**3,
    vj_train,
    vj_train**2,
    xj_train * vj_train
])


F_j_test = np.column_stack([
    np.ones_like(xj_test),
    xj_test,
    xj_test**2,
    xj_test**3,
    vj_test,
    vj_test**2,
    xj_test * vj_test
])


# multiplies every function by xj - xi
# this gives the direction to move
phi_x_i_train = (
    phi_library_train *
    (xj_train - xi_train)[:, None]
)

phi_x_i_test = (
    phi_library_test *
    (xj_test - xi_test)[:, None]
)


phi_v_i_train = (
    phi_library_train *
    (vj_train - vi_train)[:, None]
)

phi_v_i_test = (
    phi_library_test *
    (vj_test - vi_test)[:, None]
)


phi_x_j_train = (
    phi_library_train *
    (xi_train - xj_train)[:, None]
)

phi_x_j_test = (
    phi_library_test *
    (xi_test - xj_test)[:, None]
)


phi_v_j_train = (
    phi_library_train *
    (vi_train - vj_train)[:, None]
)

phi_v_j_test = (
    phi_library_test *
    (vi_test - vj_test)[:, None]
)


# full feature matrix
Theta_i_train = np.column_stack([
    F_i_train,
    phi_x_i_train,
    phi_v_i_train
])

Theta_i_test = np.column_stack([
    F_i_test,
    phi_x_i_test,
    phi_v_i_test
])


Theta_j_train = np.column_stack([
    F_j_train,
    phi_x_j_train,
    phi_v_j_train
])

Theta_j_test = np.column_stack([
    F_j_test,
    phi_x_j_test,
    phi_v_j_test
])


# theta has all of the features used to predict
# y = output (acceleration)
Theta_train = np.vstack([
    Theta_i_train,
    Theta_j_train
])

Theta_test = np.vstack([
    Theta_i_test,
    Theta_j_test
])


y_train = np.concatenate([
    vdot_i_train,
    vdot_j_train
])

y_test = np.concatenate([
    vdot_i_test,
    vdot_j_test
])


print("Training samples:", Theta_train.shape)
print("Testing samples:", Theta_test.shape)

# check feature count
print(
    "Number of training features:",
    Theta_train.shape[1]
)

if Theta_train.shape[1] != 29:

    raise ValueError(
        f"Expected 29 training features, "
        f"but found {Theta_train.shape[1]}."
    )

# standardize features
scaler = StandardScaler(
    with_mean=False
)

Theta_train_scaled = scaler.fit_transform(
    Theta_train
)

Theta_test_scaled = scaler.transform(
    Theta_test
)


# convert to PyTorch tensors
Theta_train_t = torch.tensor(
    Theta_train_scaled,
    dtype=torch.float32
)

Theta_test_t = torch.tensor(
    Theta_test_scaled,
    dtype=torch.float32
)

y_train_t = torch.tensor(
    y_train,
    dtype=torch.float32
)

y_test_t = torch.tensor(
    y_test,
    dtype=torch.float32
)


# SINDy coefficients to learn
# one coefficient per library term
coeffs = torch.zeros(
    Theta_train.shape[1],
    requires_grad=True
)


# adam optimizer
optimizer = torch.optim.Adam(
    [coeffs],
    lr=0.0001
)


# LASSO parameters
# how strongly it punishes coefficients
lambda_l1 = 0.00001

epochs = 5000


# keep track of the best model

best_loss = float("inf")
best_coeffs = None
best_epoch = 0


# training loop
for epoch in range(epochs):
    # prediction:
    # theta*c = acceleration
    prediction = (
        Theta_train_t @ coeffs
    )

    # MSE loss
    mse_loss = torch.mean(
        (prediction - y_train_t)**2
    )

    # L1 penalty for sparsity
    l1_penalty = torch.sum(
        torch.abs(coeffs[1:])
    )


    # total LASSO loss
    loss = (
        mse_loss +
        lambda_l1 * l1_penalty
    )


    # save coefficients if this
    # is the best loss so far
    if loss.item() < best_loss:

        best_loss = loss.item()

        best_coeffs = (
            coeffs.detach().clone()
        )

        best_epoch = epoch


    # clear old gradients
    optimizer.zero_grad()


    # calculate gradients
    loss.backward()


    # update coefficients
    optimizer.step()


    # updates
    if epoch % 500 == 0:

        print(
            f"Epoch {epoch}, "
            f"Loss {loss.item():.6f}, "
            f"MSE {mse_loss.item():.6f}"
        )


# restore the best coefficients
with torch.no_grad():

    coeffs.copy_(
        best_coeffs
    )


print(
    "\nBest epoch:",
    best_epoch
)

print(
    "Best loss:",
    best_loss
)

# prediction on training data
with torch.no_grad():

    train_prediction = (
        Theta_train_t @ coeffs
    )

train_prediction = train_prediction.numpy()


# prediction on test data
with torch.no_grad():

    test_prediction = (
        Theta_test_t @ coeffs
    )

test_prediction = test_prediction.numpy()


# calculate MSE
train_mse = np.mean(
    (train_prediction - y_train)**2
)

test_mse = np.mean(
    (test_prediction - y_test)**2
)
train_rmse = np.sqrt(train_mse)
test_rmse = np.sqrt(test_mse)

print("\nTraining MSE:", train_mse)
print("Test MSE:", test_mse)
print("\nTraining RMSE:", train_rmse)
print("Test RMSE:", test_rmse)

# recover learned coefficients
# detach from pytorch

learned_coeffs = coeffs.detach().numpy()


# coefficients in the original feature scale

coefficients = (
    learned_coeffs /
    scaler.scale_
)


# check coefficient count

print(
    "\nNumber of learned coefficients:",
    len(learned_coeffs)
)

if len(learned_coeffs) != 29:

    raise ValueError(
        "Number of learned coefficients is not 29."
    )


# print discovered equation

names = [
    # F(x,v)

    "1",
    "x",
    "x^2",
    "x^3",
    "v",
    "v^2",
    "x*v",

    # phi_x(r)*(xj-xi)

    "dx",
    "r dx",
    "1/(r+epsilon) dx",
    "1/(r+epsilon)^2 dx",
    "exp(-r) dx",
    "exp(-2r) dx",
    "Gaussian(0.00) dx",
    "Gaussian(0.25) dx",
    "Gaussian(0.50) dx",
    "Gaussian(0.75) dx",
    "Gaussian(1.00) dx",

    # phi_v(r)*(vj-vi)

    "dv",
    "r dv",
    "1/(r+epsilon) dv",
    "1/(r+epsilon)^2 dv",
    "exp(-r) dv",
    "exp(-2r) dv",
    "Gaussian(0.00) dv",
    "Gaussian(0.25) dv",
    "Gaussian(0.50) dv",
    "Gaussian(0.75) dv",
    "Gaussian(1.00) dv"
]


print("\nDiscovered terms")
print("----------------")

for name, c in zip(names, coefficients):

    if abs(c) > 1e-3:

        print(
            f"{c:.6f} * {name}"
        )


# print vdot data

all_vdot = np.concatenate([
    y_train,
    y_test
])


print("\n\nmean:", np.mean(all_vdot))
print("std:", np.std(all_vdot))
print("min:", np.min(all_vdot))
print("max:", np.max(all_vdot))



# prediction plot

plt.figure(figsize=(6, 6))

plt.scatter(
    y_test,
    test_prediction,
    s=5
)

plt.xlabel(
    "Measured acceleration"
)

plt.ylabel(
    "Predicted acceleration"
)

plt.title(
    "SINDy-LASSO acceleration prediction"
)

plt.grid()
plt.show()
```

```python
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# CONTINUOUS TRAJECTORY PREDICTION
#
# IMPORTANT:
# The model is initialized ONCE at t = 0.
# It then runs continuously through BOTH:
#
#   first 80%  = training region
#   final 20%  = prediction region
#
# The model is NOT restarted at the 80% boundary.
# ============================================================


# ============================================================
# 1. BUILD THETA FOR ONE CURRENT STATE
# ============================================================

def make_theta_row(
    xi_now,
    vi_now,
    xj_now,
    vj_now
):

    # --------------------------------------------------------
    # DISTANCE BETWEEN CELLS
    # --------------------------------------------------------

    r_now = abs(
        xi_now - xj_now
    )


    # --------------------------------------------------------
    # USE SAME DISTANCE NORMALIZATION AS TRAINING
    #
    # max_r was calculated ONLY from training data
    # --------------------------------------------------------

    R_scaled = (
        r_now / max_r
    )


    # --------------------------------------------------------
    # SAME INTERACTION LIBRARY AS TRAINING
    # --------------------------------------------------------

    phi = build_phi_library(
        np.array([
            R_scaled
        ])
    )[0]


    # ========================================================
    # CELL i
    # ========================================================

    # self-dynamics
    #
    # same 7 functions used during training

    F_i = np.array([
        1.0,
        xi_now,
        xi_now**2,
        xi_now**3,
        vi_now,
        vi_now**2,
        xi_now * vi_now
    ])


    # position interaction

    phi_x_i = (
        phi *
        (xj_now - xi_now)
    )


    # velocity interaction

    phi_v_i = (
        phi *
        (vj_now - vi_now)
    )


    # complete 29-term feature vector

    theta_i = np.concatenate([
        F_i,
        phi_x_i,
        phi_v_i
    ])


    # ========================================================
    # CELL j
    # ========================================================

    # self-dynamics

    F_j = np.array([
        1.0,
        xj_now,
        xj_now**2,
        xj_now**3,
        vj_now,
        vj_now**2,
        xj_now * vj_now
    ])


    # position interaction
    #
    # direction is reversed for cell j

    phi_x_j = (
        phi *
        (xi_now - xj_now)
    )


    # velocity interaction

    phi_v_j = (
        phi *
        (vi_now - vj_now)
    )


    # complete 29-term feature vector

    theta_j = np.concatenate([
        F_j,
        phi_x_j,
        phi_v_j
    ])


    return theta_i, theta_j


# ============================================================
# 2. LEARNED DIFFERENTIAL EQUATION
# ============================================================

def learned_dynamics(
    t,
    state
):

    # --------------------------------------------------------
    # CURRENT MODEL-PREDICTED STATE
    # --------------------------------------------------------

    xi_now, vi_now, xj_now, vj_now = state


    # --------------------------------------------------------
    # BUILD 29 FEATURES FROM CURRENT PREDICTED STATE
    # --------------------------------------------------------

    theta_i, theta_j = make_theta_row(
        xi_now,
        vi_now,
        xj_now,
        vj_now
    )


    # --------------------------------------------------------
    # APPLY SAME FEATURE SCALING USED DURING TRAINING
    # --------------------------------------------------------

    theta_i_scaled = scaler.transform(
        theta_i.reshape(1, -1)
    )[0]


    theta_j_scaled = scaler.transform(
        theta_j.reshape(1, -1)
    )[0]


    # --------------------------------------------------------
    # MODEL-PREDICTED ACCELERATIONS
    #
    # a_hat = Theta c
    # --------------------------------------------------------

    ai = (
        theta_i_scaled @
        learned_coeffs
    )


    aj = (
        theta_j_scaled @
        learned_coeffs
    )


    # --------------------------------------------------------
    # FIRST-ORDER ODE SYSTEM
    #
    # dxi/dt = vi
    # dvi/dt = ai
    #
    # dxj/dt = vj
    # dvj/dt = aj
    # --------------------------------------------------------

    return [
        vi_now,
        ai,
        vj_now,
        aj
    ]


# ============================================================
# 3. FULL TIME AXIS
# ============================================================

num_points = len(xi)


# IMPORTANT:
#
# Your velocities and accelerations were calculated using
# dt = 10 minutes.
#
# Therefore solve_ivp must ALSO integrate in minutes.
#
# We will convert the x-axis to 0-100% only for the plot.

time_full = (
    np.arange(num_points) *
    dt
)


print(
    "Number of trajectory points:",
    num_points
)

print(
    "Train/test split index:",
    split
)

print(
    "Training fraction:",
    split / num_points
)


# ============================================================
# 4. INITIAL CONDITION
#
# INITIALIZE THE MODEL EXACTLY ONCE
#
# t = 0
# ============================================================

initial_state = [
    xi[0],
    vi[0],
    xj[0],
    vj[0]
]


print(
    "\nInitial state:"
)

print(
    "Cell i position:",
    xi[0]
)

print(
    "Cell i velocity:",
    vi[0]
)

print(
    "Cell j position:",
    xj[0]
)

print(
    "Cell j velocity:",
    vj[0]
)


# ============================================================
# 5. CONTINUOUS FORWARD SIMULATION
#
# ONE solve_ivp CALL
#
# NO RESTART AT 80%
# ============================================================

solution_full = solve_ivp(

    learned_dynamics,

    (
        time_full[0],
        time_full[-1]
    ),

    initial_state,

    t_eval=time_full,

    method="RK45",

    rtol=1e-6,

    atol=1e-8,

    max_step=dt / 5
)


# ============================================================
# 6. CHECK SOLVER
# ============================================================

print(
    "\nSimulation successful:",
    solution_full.success
)

print(
    "Solver message:",
    solution_full.message
)


if not solution_full.success:

    raise RuntimeError(
        "Continuous trajectory simulation failed: "
        +
        solution_full.message
    )


# ============================================================
# 7. EXTRACT FULL CONTINUOUS PREDICTION
# ============================================================

xi_pred_full = (
    solution_full.y[0]
)

vi_pred_full = (
    solution_full.y[1]
)

xj_pred_full = (
    solution_full.y[2]
)

vj_pred_full = (
    solution_full.y[3]
)


print(
    "\nPredicted points:",
    len(xi_pred_full)
)


# ============================================================
# 8. SPLIT THE ALREADY-COMPLETED SIMULATION
#
# THIS DOES NOT RESTART THE MODEL.
#
# It only separates the ONE continuous prediction afterward
# so that we can calculate training and testing errors.
# ============================================================

xi_pred_train = (
    xi_pred_full[:split]
)

xi_pred_test = (
    xi_pred_full[split:]
)


xj_pred_train = (
    xj_pred_full[:split]
)

xj_pred_test = (
    xj_pred_full[split:]
)


# ============================================================
# 9. TRAINING POSITION RMSE
# ============================================================

train_position_rmse_i = np.sqrt(
    np.mean(
        (
            xi_pred_train -
            xi_train
        )**2
    )
)


train_position_rmse_j = np.sqrt(
    np.mean(
        (
            xj_pred_train -
            xj_train
        )**2
    )
)


train_position_rmse_overall = np.sqrt(
    (
        np.mean(
            (
                xi_pred_train -
                xi_train
            )**2
        )
        +
        np.mean(
            (
                xj_pred_train -
                xj_train
            )**2
        )
    )
    / 2
)


# ============================================================
# 10. TESTING POSITION RMSE
#
# IMPORTANT:
#
# These test predictions come from the SAME simulation
# initialized at t = 0.
#
# No observed test position or velocity was supplied
# to the model at the 80% boundary.
# ============================================================

test_position_rmse_i = np.sqrt(
    np.mean(
        (
            xi_pred_test -
            xi_test
        )**2
    )
)


test_position_rmse_j = np.sqrt(
    np.mean(
        (
            xj_pred_test -
            xj_test
        )**2
    )
)


test_position_rmse_overall = np.sqrt(
    (
        np.mean(
            (
                xi_pred_test -
                xi_test
            )**2
        )
        +
        np.mean(
            (
                xj_pred_test -
                xj_test
            )**2
        )
    )
    / 2
)


# ============================================================
# 11. PRINT POSITION ERRORS
# ============================================================

print(
    "\n===================================="
)

print(
    "CONTINUOUS POSITION PREDICTION RMSE"
)

print(
    "===================================="
)


print(
    "\nTRAINING REGION"
)

print(
    "Cell i:",
    train_position_rmse_i
)

print(
    "Cell j:",
    train_position_rmse_j
)

print(
    "Overall:",
    train_position_rmse_overall
)


print(
    "\nTEST REGION"
)

print(
    "Cell i:",
    test_position_rmse_i
)

print(
    "Cell j:",
    test_position_rmse_j
)

print(
    "Overall:",
    test_position_rmse_overall
)


# ============================================================
# 12. CREATE 0-100 TRAJECTORY AXIS
#
# This is ONLY for displaying the trajectory.
#
# solve_ivp above was correctly run in minutes.
# ============================================================

time_percent = np.linspace(
    0,
    100,
    num_points
)


# exact location of training boundary on percentage axis

split_percent = (
    100 *
    split /
    num_points
)


print(
    "\nTrain/test boundary on plot:",
    split_percent
)


# ============================================================
# 13. FINAL TRAJECTORY PLOT
# ============================================================

plt.figure(
    figsize=(10, 5)
)


# ============================================================
# CELL i
# ============================================================

plt.plot(
    time_percent,
    xi,
    linewidth=2,
    label=f"Cell {i} observed"
)


plt.plot(
    time_percent,
    xi_pred_full,
    "--",
    linewidth=2.5,
    label=f"Cell {i} model"
)


# ============================================================
# CELL j
# ============================================================

plt.plot(
    time_percent,
    xj,
    linewidth=2,
    label=f"Cell {j} observed"
)


plt.plot(
    time_percent,
    xj_pred_full,
    "--",
    linewidth=2.5,
    label=f"Cell {j} model"
)


# ============================================================
# TRAIN / TEST BOUNDARY
#
# This line is ONLY a visual marker.
#
# Nothing happens to the simulation here.
# ============================================================

plt.axvline(
    x=split_percent,
    linestyle=":",
    linewidth=2,
    label="Train/Test boundary"
)


# ============================================================
# OPTIONAL TRAIN / TEST LABELS
# ============================================================

y_min, y_max = plt.ylim()

label_y = (
    y_max -
    0.05 * (y_max - y_min)
)


plt.text(
    split_percent / 2,
    label_y,
    "Training",
    horizontalalignment="center"
)


plt.text(
    (
        split_percent +
        100
    ) / 2,
    label_y,
    "Prediction",
    horizontalalignment="center"
)


# ============================================================
# FINAL FORMATTING
# ============================================================

plt.xlim(
    0,
    100
)


plt.xlabel(
    "Trajectory (%)"
)


plt.ylabel(
    "Position"
)


plt.title(
    f"Continuous Trajectory Prediction: "
    f"Cell Pair ({i}, {j})"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.show()
```

#BASIC SINDY LIBRARY

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from scipy.signal import savgol_filter
from scipy.integrate import solve_ivp

import torch


np.random.seed(42)


# load data

data = np.genfromtxt("pnas.2016602118.sd02.txt")
print("Raw data shape:", data.shape)


# parameters

dt = 10.0
num_cells = data.shape[0]


# shared interaction feature library

# this SAME function is used during both
# training and trajectory prediction

def build_phi_library(R_scaled):

    phi_library = np.column_stack([
        np.ones_like(R_scaled),
        R_scaled,
        R_scaled**2
    ])

    return phi_library


# cells are already paired:

# (0,1)

# (2,3)

# (4,5)

# make pairs

pairs = []

for i in range(0, num_cells - 1, 2):
    pairs.append((i, i + 1))

print("Number of interacting pairs:", len(pairs))


# choose ONE interacting pair

pair_index = 23

i, j = pairs[pair_index]

print("Using pair:", (i, j))


# trajectory of cell i

xi = data[i]


# trajectory for cell j

xj = data[j]


# remove NaNs

xi = xi[~np.isnan(xi)]
xj = xj[~np.isnan(xj)]


# get shorter length, make both equal

length = min(len(xi), len(xj))


# need enough points for Savitzky-Golay filtering

if length < 21:

    raise ValueError(
        "Not enough trajectory points for Savitzky-Golay filtering."
    )


# fix lengths to shorter one so both are same length

xi = xi[:length]
xj = xj[:length]


# smooth positions

xi_smooth = savgol_filter(
    xi,
    window_length=21,
    polyorder=3
)

xj_smooth = savgol_filter(
    xj,
    window_length=21,
    polyorder=3
)


# calculate velocity directly using Savitzky-Golay derivative

vi = savgol_filter(
    xi,
    window_length=21,
    polyorder=3,
    deriv=1,
    delta=dt
)

vj = savgol_filter(
    xj,
    window_length=21,
    polyorder=3,
    deriv=1,
    delta=dt
)


# calculate acceleration directly using Savitzky-Golay derivative

vdot_i = savgol_filter(
    xi,
    window_length=21,
    polyorder=3,
    deriv=2,
    delta=dt
)

vdot_j = savgol_filter(
    xj,
    window_length=21,
    polyorder=3,
    deriv=2,
    delta=dt
)


# use smoothed positions

xi = xi_smooth
xj = xj_smooth


# get distance

r = np.abs(xi - xj)


# split trajectory into first 80% and final 20%

split = int(0.8 * len(r))

print("Training trajectory points:", split)
print("Testing trajectory points:", len(r) - split)


# split positions, velocities, distances, and acceleration

xi_train = xi[:split]
xi_test = xi[split:]

xj_train = xj[:split]
xj_test = xj[split:]

vi_train = vi[:split]
vi_test = vi[split:]

vj_train = vj[:split]
vj_test = vj[split:]

vdot_i_train = vdot_i[:split]
vdot_i_test = vdot_i[split:]

vdot_j_train = vdot_j[:split]
vdot_j_test = vdot_j[split:]

r_train = r[:split]
r_test = r[split:]


# normalize distance for this pair

max_r = np.max(r_train)

if max_r == 0:

    raise ValueError(
        "Maximum training distance is zero."
    )


R_train_scaled = r_train / max_r
R_test_scaled = r_test / max_r


# interaction library

# this is the SAME library used later
# during trajectory prediction

phi_library_train = build_phi_library(
    R_train_scaled
)

phi_library_test = build_phi_library(
    R_test_scaled
)


# SINDy library for F(x, v) cell i

F_i_train = np.column_stack([
    np.ones_like(xi_train),
    xi_train,
    vi_train
])


F_i_test = np.column_stack([
    np.ones_like(xi_test),
    xi_test,
    vi_test
])


# SINDy library for F(x, v) cell j

F_j_train = np.column_stack([
    np.ones_like(xj_train),
    xj_train,
    vj_train
])


F_j_test = np.column_stack([
    np.ones_like(xj_test),
    xj_test,
    vj_test
])


# multiplies every function by xj - xi
# this gives the direction to move

phi_x_i_train = (
    phi_library_train *
    (xj_train - xi_train)[:, None]
)

phi_x_i_test = (
    phi_library_test *
    (xj_test - xi_test)[:, None]
)


phi_v_i_train = (
    phi_library_train *
    (vj_train - vi_train)[:, None]
)

phi_v_i_test = (
    phi_library_test *
    (vj_test - vi_test)[:, None]
)


phi_x_j_train = (
    phi_library_train *
    (xi_train - xj_train)[:, None]
)

phi_x_j_test = (
    phi_library_test *
    (xi_test - xj_test)[:, None]
)


phi_v_j_train = (
    phi_library_train *
    (vi_train - vj_train)[:, None]
)

phi_v_j_test = (
    phi_library_test *
    (vi_test - vj_test)[:, None]
)


# full feature matrix

Theta_i_train = np.column_stack([
    F_i_train,
    phi_x_i_train,
    phi_v_i_train
])

Theta_i_test = np.column_stack([
    F_i_test,
    phi_x_i_test,
    phi_v_i_test
])


Theta_j_train = np.column_stack([
    F_j_train,
    phi_x_j_train,
    phi_v_j_train
])

Theta_j_test = np.column_stack([
    F_j_test,
    phi_x_j_test,
    phi_v_j_test
])


# theta has all of the features used to predict
# y = output (acceleration)

Theta_train = np.vstack([
    Theta_i_train,
    Theta_j_train
])

Theta_test = np.vstack([
    Theta_i_test,
    Theta_j_test
])


y_train = np.concatenate([
    vdot_i_train,
    vdot_j_train
])

y_test = np.concatenate([
    vdot_i_test,
    vdot_j_test
])


print("Training samples:", Theta_train.shape)
print("Testing samples:", Theta_test.shape)


# check feature count

print(
    "Number of training features:",
    Theta_train.shape[1]
)


if Theta_train.shape[1] != 9:

    raise ValueError(
        f"Expected 9 training features, "
        f"but found {Theta_train.shape[1]}."
    )


# standardize features

scaler = StandardScaler(
    with_mean=False
)


Theta_train_scaled = scaler.fit_transform(
    Theta_train
)

Theta_test_scaled = scaler.transform(
    Theta_test
)


# convert to PyTorch tensors

Theta_train_t = torch.tensor(
    Theta_train_scaled,
    dtype=torch.float32
)

Theta_test_t = torch.tensor(
    Theta_test_scaled,
    dtype=torch.float32
)

y_train_t = torch.tensor(
    y_train,
    dtype=torch.float32
)

y_test_t = torch.tensor(
    y_test,
    dtype=torch.float32
)


# SINDy coefficients to learn
# one coefficient per library term

coeffs = torch.zeros(
    Theta_train.shape[1],
    requires_grad=True
)


# adam optimizer

optimizer = torch.optim.Adam(
    [coeffs],
    lr=0.0001
)


# LASSO parameters
# how strongly it punishes coefficients

lambda_l1 = 0.00001

epochs = 5000


# keep track of the best model

best_loss = float("inf")
best_coeffs = None
best_epoch = 0


# training loop

for epoch in range(epochs):

    # prediction:
    # theta*c = acceleration

    prediction = (
        Theta_train_t @ coeffs
    )


    # MSE loss

    mse_loss = torch.mean(
        (prediction - y_train_t)**2
    )


    # L1 penalty for sparsity

    l1_penalty = torch.sum(
        torch.abs(coeffs[1:])
    )


    # total LASSO loss

    loss = (
        mse_loss +
        lambda_l1 * l1_penalty
    )


    # save coefficients if this
    # is the best loss so far

    if loss.item() < best_loss:

        best_loss = loss.item()

        best_coeffs = (
            coeffs.detach().clone()
        )

        best_epoch = epoch


    # clear old gradients

    optimizer.zero_grad()


    # calculate gradients

    loss.backward()


    # update coefficients

    optimizer.step()


    # updates

    if epoch % 500 == 0:

        print(
            f"Epoch {epoch}, "
            f"Loss {loss.item():.6f}, "
            f"MSE {mse_loss.item():.6f}"
        )


# restore the best coefficients

with torch.no_grad():

    coeffs.copy_(
        best_coeffs
    )


print(
    "\nBest epoch:",
    best_epoch
)


print(
    "Best loss:",
    best_loss
)


# prediction on training data

with torch.no_grad():

    train_prediction = (
        Theta_train_t @ coeffs
    )


train_prediction = train_prediction.numpy()


# prediction on test data

with torch.no_grad():

    test_prediction = (
        Theta_test_t @ coeffs
    )


test_prediction = test_prediction.numpy()


# calculate MSE

train_mse = np.mean(
    (train_prediction - y_train)**2
)

test_mse = np.mean(
    (test_prediction - y_test)**2
)


train_rmse = np.sqrt(
    train_mse
)

test_rmse = np.sqrt(
    test_mse
)


print(
    "\nTraining MSE:",
    train_mse
)

print(
    "Test MSE:",
    test_mse
)

print(
    "\nTraining RMSE:",
    train_rmse
)

print(
    "Test RMSE:",
    test_rmse
)


# recover learned coefficients
# detach from pytorch

learned_coeffs = coeffs.detach().numpy()


# coefficients in the original feature scale

coefficients = (
    learned_coeffs /
    scaler.scale_
)


# check coefficient count

print(
    "\nNumber of learned coefficients:",
    len(learned_coeffs)
)


if len(learned_coeffs) != 9:

    raise ValueError(
        "Number of learned coefficients is not 9."
    )


# print discovered equation

names = [

    # F(x,v)

    "1",
    "x",
    "v",

    # phi_x(r)*(xj-xi)

    "dx",
    "r dx",
    "r^2 dx",

    # phi_v(r)*(vj-vi)

    "dv",
    "r dv",
    "r^2 dv"
]


print(
    "\nDiscovered terms"
)

print(
    "----------------"
)


for name, c in zip(
    names,
    coefficients
):

    if abs(c) > 1e-3:

        print(
            f"{c:.6f} * {name}"
        )


# print vdot data

all_vdot = np.concatenate([
    y_train,
    y_test
])


print(
    "\n\nmean:",
    np.mean(all_vdot)
)

print(
    "std:",
    np.std(all_vdot)
)

print(
    "min:",
    np.min(all_vdot)
)

print(
    "max:",
    np.max(all_vdot)
)


# prediction plot

plt.figure(
    figsize=(6, 6)
)


plt.scatter(
    y_test,
    test_prediction,
    s=5
)


plt.xlabel(
    "Measured acceleration"
)

plt.ylabel(
    "Predicted acceleration"
)

plt.title(
    "SINDy-LASSO acceleration prediction"
)

plt.grid()

plt.show()


# ============================================================
# CREATE PREDICTED POSITION TRAJECTORIES
# ============================================================


# number of points in held-out test region

num_test_points = len(xi_test)


# time begins at 0 for the test simulation

time_test = np.arange(
    num_test_points
) * dt


# build one row of Theta for a particular state

def make_theta_row(
    xi_now,
    vi_now,
    xj_now,
    vj_now
):

    # distance between cells

    r_now = abs(
        xi_now - xj_now
    )


    # use SAME distance normalization as training

    R_scaled = (
        r_now / max_r
    )


    # interaction library

    phi = build_phi_library(
        np.array([
            R_scaled
        ])
    )[0]


    # Cell i

    F_i = np.array([
        1.0,
        xi_now,
        vi_now
    ])


    phi_x_i = (
        phi *
        (xj_now - xi_now)
    )


    phi_v_i = (
        phi *
        (vj_now - vi_now)
    )


    theta_i = np.concatenate([
        F_i,
        phi_x_i,
        phi_v_i
    ])


    # Cell j

    F_j = np.array([
        1.0,
        xj_now,
        vj_now
    ])


    phi_x_j = (
        phi *
        (xi_now - xj_now)
    )


    phi_v_j = (
        phi *
        (vi_now - vj_now)
    )


    theta_j = np.concatenate([
        F_j,
        phi_x_j,
        phi_v_j
    ])


    return theta_i, theta_j


# ODE learned by the model

def learned_dynamics(
    t,
    state
):

    xi_now, vi_now, xj_now, vj_now = state


    theta_i, theta_j = make_theta_row(
        xi_now,
        vi_now,
        xj_now,
        vj_now
    )


    # apply SAME feature scaling used during training

    theta_i_scaled = scaler.transform(
        theta_i.reshape(1, -1)
    )[0]


    theta_j_scaled = scaler.transform(
        theta_j.reshape(1, -1)
    )[0]


    # predicted acceleration

    ai = (
        theta_i_scaled @
        learned_coeffs
    )

    aj = (
        theta_j_scaled @
        learned_coeffs
    )


    # state derivatives:
    #
    # dxi/dt = vi
    # dvi/dt = ai
    # dxj/dt = vj
    # dvj/dt = aj

    return [
        vi_now,
        ai,
        vj_now,
        aj
    ]


# Initial condition
# start prediction exactly where testing begins

initial_state = [
    xi[split],
    vi[split],
    xj[split],
    vj[split]
]


# solve the ODE

solution = solve_ivp(
    learned_dynamics,
    [
        time_test[0],
        time_test[-1]
    ],
    initial_state,
    t_eval=time_test,
    method="RK45",
    rtol=1e-6,
    atol=1e-8,
    max_step=dt / 5
)


# EXTRACT POSITION PREDICTIONS

xi_pred = solution.y[0]
xj_pred = solution.y[2]


print(
    "xi_pred shape:",
    xi_pred.shape
)

print(
    "xj_pred shape:",
    xj_pred.shape
)


# position RMSE

position_rmse_i = np.sqrt(
    np.mean(
        (xi_pred - xi_test)**2
    )
)


position_rmse_j = np.sqrt(
    np.mean(
        (xj_pred - xj_test)**2
    )
)


position_rmse_overall = np.sqrt(
    (
        np.mean(
            (xi_pred - xi_test)**2
        )
        +
        np.mean(
            (xj_pred - xj_test)**2
        )
    ) / 2
)


print(
    "\nCell i position RMSE:",
    position_rmse_i
)

print(
    "Cell j position RMSE:",
    position_rmse_j
)

print(
    "Overall position RMSE:",
    position_rmse_overall
)
```

```python
# ============================================================
# CREATE TRAINING + TESTING POSITION PREDICTIONS
# ============================================================


# ============================================================
# BUILD THETA FOR ONE CURRENT STATE
# ============================================================

def make_theta_row(
    xi_now,
    vi_now,
    xj_now,
    vj_now
):

    # ------------------------------------------
    # distance between cells
    # ------------------------------------------

    r_now = abs(
        xi_now - xj_now
    )


    # ------------------------------------------
    # use SAME distance normalization
    # as training
    # ------------------------------------------

    R_scaled = (
        r_now / max_r
    )


    # ------------------------------------------
    # interaction library
    # ------------------------------------------

    phi = build_phi_library(
        np.array([
            R_scaled
        ])
    )[0]


    # ========================================================
    # CELL i
    # ========================================================


    # self-dynamics library
    # 3 terms: 1, x, v

    F_i = np.array([
        1.0,
        xi_now,
        vi_now
    ])


    # position interaction
    # 3 terms: dx, r*dx, r^2*dx

    phi_x_i = (
        phi *
        (xj_now - xi_now)
    )


    # velocity interaction
    # 3 terms: dv, r*dv, r^2*dv

    phi_v_i = (
        phi *
        (vj_now - vi_now)
    )


    # all 9 features for cell i

    theta_i = np.concatenate([
        F_i,
        phi_x_i,
        phi_v_i
    ])


    # ========================================================
    # CELL j
    # ========================================================


    # self-dynamics library
    # 3 terms: 1, x, v

    F_j = np.array([
        1.0,
        xj_now,
        vj_now
    ])


    # position interaction
    # direction reverses for cell j

    phi_x_j = (
        phi *
        (xi_now - xj_now)
    )


    # velocity interaction
    # direction reverses for cell j

    phi_v_j = (
        phi *
        (vi_now - vj_now)
    )


    # all 9 features for cell j

    theta_j = np.concatenate([
        F_j,
        phi_x_j,
        phi_v_j
    ])


    return theta_i, theta_j


# ============================================================
# ODE LEARNED BY THE MODEL
# ============================================================

def learned_dynamics(
    t,
    state
):

    # current state

    xi_now, vi_now, xj_now, vj_now = state


    # ------------------------------------------
    # build current features
    # ------------------------------------------

    theta_i, theta_j = make_theta_row(
        xi_now,
        vi_now,
        xj_now,
        vj_now
    )


    # ------------------------------------------
    # apply SAME scaling used during training
    # ------------------------------------------

    theta_i_scaled = scaler.transform(
        theta_i.reshape(1, -1)
    )[0]


    theta_j_scaled = scaler.transform(
        theta_j.reshape(1, -1)
    )[0]


    # ------------------------------------------
    # predict acceleration
    # ------------------------------------------

    ai = (
        theta_i_scaled @
        learned_coeffs
    )


    aj = (
        theta_j_scaled @
        learned_coeffs
    )


    # ------------------------------------------
    # state derivatives
    #
    # dxi/dt = vi
    # dvi/dt = ai
    #
    # dxj/dt = vj
    # dvj/dt = aj
    # ------------------------------------------

    return [
        vi_now,
        ai,
        vj_now,
        aj
    ]


# ============================================================
# TRAINING PREDICTION
# ============================================================


num_train_points = len(
    xi_train
)


# training simulation time

time_train = np.arange(
    num_train_points
) * dt


# ------------------------------------------
# start at first training point
# ------------------------------------------

initial_state_train = [
    xi[0],
    vi[0],
    xj[0],
    vj[0]
]


# ------------------------------------------
# simulate through training region
# ------------------------------------------

solution_train = solve_ivp(
    learned_dynamics,
    [
        time_train[0],
        time_train[-1]
    ],
    initial_state_train,
    t_eval=time_train,
    method="RK45",
    rtol=1e-6,
    atol=1e-8,
    max_step=dt / 5
)


# ------------------------------------------
# check solver
# ------------------------------------------

print(
    "Training simulation successful:",
    solution_train.success
)


# ------------------------------------------
# extract training predictions
# ------------------------------------------

xi_pred_train = (
    solution_train.y[0]
)

vi_pred_train = (
    solution_train.y[1]
)

xj_pred_train = (
    solution_train.y[2]
)

vj_pred_train = (
    solution_train.y[3]
)


print(
    "xi_pred_train shape:",
    xi_pred_train.shape
)

print(
    "xj_pred_train shape:",
    xj_pred_train.shape
)


# ============================================================
# TESTING PREDICTION
# ============================================================


num_test_points = len(
    xi_test
)


# testing simulation begins at 0
# relative to train/test boundary

time_test = np.arange(
    num_test_points
) * dt


# ------------------------------------------
# start exactly where testing begins
# ------------------------------------------

initial_state_test = [
    xi[split],
    vi[split],
    xj[split],
    vj[split]
]


# ------------------------------------------
# simulate through testing region
# ------------------------------------------

solution_test = solve_ivp(
    learned_dynamics,
    [
        time_test[0],
        time_test[-1]
    ],
    initial_state_test,
    t_eval=time_test,
    method="RK45",
    rtol=1e-6,
    atol=1e-8,
    max_step=dt / 5
)


# ------------------------------------------
# check solver
# ------------------------------------------

print(
    "Testing simulation successful:",
    solution_test.success
)


# ------------------------------------------
# extract testing predictions
# ------------------------------------------

xi_pred_test = (
    solution_test.y[0]
)

vi_pred_test = (
    solution_test.y[1]
)

xj_pred_test = (
    solution_test.y[2]
)

vj_pred_test = (
    solution_test.y[3]
)


print(
    "xi_pred_test shape:",
    xi_pred_test.shape
)

print(
    "xj_pred_test shape:",
    xj_pred_test.shape
)


# ============================================================
# POSITION RMSE
# ============================================================


# ------------------------------------------
# TRAINING POSITION RMSE
# ------------------------------------------

train_position_rmse_i = np.sqrt(
    np.mean(
        (
            xi_pred_train -
            xi_train
        )**2
    )
)


train_position_rmse_j = np.sqrt(
    np.mean(
        (
            xj_pred_train -
            xj_train
        )**2
    )
)


train_position_rmse_overall = np.sqrt(
    (
        np.mean(
            (
                xi_pred_train -
                xi_train
            )**2
        )
        +
        np.mean(
            (
                xj_pred_train -
                xj_train
            )**2
        )
    ) / 2
)


# ------------------------------------------
# TESTING POSITION RMSE
# ------------------------------------------

test_position_rmse_i = np.sqrt(
    np.mean(
        (
            xi_pred_test -
            xi_test
        )**2
    )
)


test_position_rmse_j = np.sqrt(
    np.mean(
        (
            xj_pred_test -
            xj_test
        )**2
    )
)


test_position_rmse_overall = np.sqrt(
    (
        np.mean(
            (
                xi_pred_test -
                xi_test
            )**2
        )
        +
        np.mean(
            (
                xj_pred_test -
                xj_test
            )**2
        )
    ) / 2
)


# ------------------------------------------
# print RMSE
# ------------------------------------------

print(
    "\nCell i Training Position RMSE:",
    train_position_rmse_i
)

print(
    "Cell j Training Position RMSE:",
    train_position_rmse_j
)

print(
    "Overall Training Position RMSE:",
    train_position_rmse_overall
)


print(
    "\nCell i Testing Position RMSE:",
    test_position_rmse_i
)

print(
    "Cell j Testing Position RMSE:",
    test_position_rmse_j
)

print(
    "Overall Testing Position RMSE:",
    test_position_rmse_overall
)


# ============================================================
# FINAL POSTER PLOT:
# RAW DATA + TRAINING + TESTING PREDICTIONS
# ============================================================


# ------------------------------------------
# get ORIGINAL raw trajectories again
# ------------------------------------------

xi_raw_plot = data[i]
xj_raw_plot = data[j]


# ------------------------------------------
# remove NaNs exactly like preprocessing
# ------------------------------------------

xi_raw_plot = xi_raw_plot[
    ~np.isnan(
        xi_raw_plot
    )
]

xj_raw_plot = xj_raw_plot[
    ~np.isnan(
        xj_raw_plot
    )
]


# ------------------------------------------
# make them the same length
# ------------------------------------------

length = min(
    len(xi_raw_plot),
    len(xj_raw_plot)
)


xi_raw_plot = (
    xi_raw_plot[:length]
)

xj_raw_plot = (
    xj_raw_plot[:length]
)


# ============================================================
# TIME AXES
# ============================================================


# full raw-data time axis

time_raw = np.arange(
    length
) * dt


# ------------------------------------------
# training prediction time
# ------------------------------------------

time_pred_train = np.arange(
    len(xi_pred_train)
) * dt


# ------------------------------------------
# testing prediction starts
# at train/test split
# ------------------------------------------

time_pred_test = (
    np.arange(
        len(xi_pred_test)
    )
    +
    split
) * dt


# ============================================================
# PLOT
# ============================================================

plt.figure(
    figsize=(10, 5)
)


# ============================================================
# CELL i
# ============================================================


# observed

plt.plot(
    time_raw,
    xi_raw_plot,
    alpha=0.7,
    label=f"Cell {i} observed"
)


# training prediction

plt.plot(
    time_pred_train,
    xi_pred_train,
    "--",
    linewidth=2.5,
    label=f"Cell {i} training prediction"
)


# testing prediction

plt.plot(
    time_pred_test,
    xi_pred_test,
    "--",
    linewidth=2.5,
    label=f"Cell {i} testing prediction"
)


# ============================================================
# CELL j
# ============================================================


# observed

plt.plot(
    time_raw,
    xj_raw_plot,
    alpha=0.7,
    label=f"Cell {j} observed"
)


# training prediction

plt.plot(
    time_pred_train,
    xj_pred_train,
    "--",
    linewidth=2.5,
    label=f"Cell {j} training prediction"
)


# testing prediction

plt.plot(
    time_pred_test,
    xj_pred_test,
    "--",
    linewidth=2.5,
    label=f"Cell {j} testing prediction"
)


# ============================================================
# TRAIN / TEST BOUNDARY
# ============================================================

plt.axvline(
    split * dt,
    linestyle=":",
    label="Train/Test split"
)


# ============================================================
# LABELS
# ============================================================

plt.xlabel(
    "Time (min)"
)

plt.ylabel(
    "Position"
)


plt.title(
    f"Observed and Predicted Cell Trajectories for Basic SINDy Model (9 Candidate Functions): Pair ({i}, {j})"
)


plt.legend()

plt.grid()

plt.tight_layout()

plt.show()
```
